## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Berapa persen pertumbuhan total penyewaan sepeda tahun 2012 dibanding 2011, dan di bulan apa pertumbuhannya paling tinggi, agar operator bisa merencanakan penambahan sesuai bulan permintaan naik?
- **Pertanyaan 2:** Pada rentang suhu berapa (per 5°C) rata-rata penyewaan harian tertinggi selama 2011-2012, dan di suhu berapa penyewaan turun lebih dari 25% dari puncaknya, agar operator bisa menetapkan ambang suhu untuk menambah atau mengurangi sepeda?
- **Pertanyaan 3:** Berapa persentase penyewaan oleh pengguna casual dan registered terhadap total penyewaan sepeda selama 2011-2012, agar operator bisa menentukan segmen yang diprioritaskan untuk promosi?

## Exploratory Data Analysis (EDA)

### Explore untuk Pertanyaan Pertama

```python
# Menghitung total penyewaan sepeda per tahun
yearly_rentals = df.groupby('yr')['cnt'].sum()

yearly_rentals
```

Dari data tersebut diketahui bahwa jumlah penyewaan selama tahun 2011 - 2012

```python
# Menghitung persentase pertumbuhan total penyewaan
growth_2012 = (
    (yearly_rentals[2012] - yearly_rentals[2011])
    / yearly_rentals[2011]
) * 100

print(f"Pertumbuhan total penyewaan tahun 2012 dibanding 2011: {growth_2012:.2f}%")
```

Jadi terdapat pertumbuhan yang tinggi sebanyak 64.88% dari tahun 2011 ke 2012

```python
# Menghitung total penyewaan per bulan untuk setiap tahun
monthly_rentals = df.groupby(['mnth', 'yr'])['cnt'].sum().unstack()

# Mengubah kode tahun menjadi nama kolom
monthly_rentals.columns = [2011, 2012]

# Menghitung persentase pertumbuhan setiap bulan
monthly_rentals['growth_pct'] = (
    (monthly_rentals[2012] - monthly_rentals[2011])
    / monthly_rentals[2011]
) * 100

monthly_rentals.sort_values('growth_pct', ascending=False)
```

**Insight:** (Opsional)
- Total penyewaan sepeda meningkat dari 1.243.103 pada tahun 2011 menjadi 2.049.576 pada tahun 2012, dengan pertumbuhan sebesar 64,88%.
- Pertumbuhan penyewaan tertinggi terjadi pada bulan Maret, yaitu sebesar 157,44% dibandingkan bulan Maret tahun 2011.

### Explore untuk Pertanyaan Kedua

```python
# Membuat rentang suhu setiap 5°C
bins = range(0, 41, 5)
labels = [f"{i}-{i+5}°C" for i in range(0, 40, 5)]

df['temp_range'] = pd.cut(
    df['temp_c'],
    bins=bins,
    labels=labels,
    right=False
)

# Menghitung rata-rata penyewaan harian pada setiap rentang suhu
avg_rentals_by_temp = df.groupby(
    'temp_range',
    observed=False
)['cnt'].mean()

avg_rentals_by_temp.sort_values(ascending=False)
```

Jadi puncak rata-rata penyewaan ada pada rentang 25–30°C, dengan rata-rata sekitar 5.726 penyewaan per hari.

```python
# Menentukan nilai puncak rata-rata penyewaan
peak_rentals = avg_rentals_by_temp.max()

# Menentukan batas 25% di bawah puncak
threshold_25 = peak_rentals * 0.75

print(f"Puncak rata-rata penyewaan: {peak_rentals:.2f}")
print(f"Batas penurunan 25%: {threshold_25:.2f}")

# Mencari rentang suhu dengan rata-rata penyewaan
# yang turun lebih dari 25% dari puncak
temp_drop_25 = avg_rentals_by_temp[
    avg_rentals_by_temp < threshold_25
]

temp_drop_25.sort_values(ascending=False)
```

Untuk mencari nilai penuruan sebanyak 25% darai puncaknya, cukup mengalikan nilai puncaknya dengan 0.75. Setelah itu didapat bahwa 4 rentang suhu yang mengalami penurunan sebanyak 25% dari puncaknya.

**Insight** (Opsional)
- Rata-rata penyewaan sepeda tertinggi terjadi pada rentang suhu 25–30°C, yaitu sebesar 5.726,24 penyewaan per hari.
- Penyewaan sepeda turun lebih dari 25% dari nilai puncaknya pada rentang suhu 15–20°C, dengan rata-rata 4.191,98 penyewaan per hari. Hal ini menunjukkan bahwa suhu di bawah kisaran optimal 25–30°C berkaitan dengan penurunan jumlah penyewaan yang cukup besar.

### Explore untuk Pertanyaan Ketiga

```python
# Menghitung total penyewaan berdasarkan tipe pengguna
total_casual = df['casual'].sum()
total_registered = df['registered'].sum()
total_rentals = df['cnt'].sum()

# Menghitung persentase masing-masing tipe pengguna
casual_pct = (total_casual / total_rentals) * 100
registered_pct = (total_registered / total_rentals) * 100

print(f"Persentase penyewaan pengguna casual: {casual_pct:.2f}%")
print(f"Persentase penyewaan pengguna registered: {registered_pct:.2f}%")
```

Persentase penyewaan didominasi oleh pengguna registered sebanyak 81.17%

**Insight** (Opsional)
- Pengguna registered menyumbang sebesar 81.17% dari total penyewaan sepeda selama 2011–2012, sedangkan pengguna casual menyumbang sebesar 18.83%.
- Hal ini menunjukkan bahwa penyewaan sepeda selama periode tersebut lebih banyak berasal dari pengguna registered
